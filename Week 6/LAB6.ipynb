### Lab 6.1 : Simple Calculator Functions

```python
def add(a, b):
    return a+b

def substract(a, b):
    return a-b

def multiply(a, b):
    return a*b

def divide(a, b):
    return a/b

def power(a, b=2):
    return a**b

print("========== Welcome to Calculator ==========")
while True:
    try:
        print("-------------------------------------------")
        print("Please choose an operation below:")
        print("1. add")
        print("2. substract")
        print("3. multiply")
        print("4. divide")
        print("5. power")
        print("6. exit")
        operator = int(input("Choose one (1 - 6): ").strip())

        if(operator > 6) or (operator < 1):
            print("Please enter the number beetween 1 - 6")
        elif operator == 6:
            print("-------------------------------------------")
            print("Thank you...")
            break
        elif operator == 1:
            print("-------------------------------------------")
            print("you choose 1: Addition")
            a = float(input("Enter the 1st addend: ").strip())
            b = float(input("Enter the 2nd addend: ").strip())
            result = add(a, b)
        elif operator == 2:
            print("-------------------------------------------")
            print("you choose 2: Substraction")
            a = float(input("Enter the minuend: ").strip())
            b = float(input("Enter the subtrahend: ").strip())
            result = substract(a, b)
        elif operator == 3:
            print("-------------------------------------------")
            print("you choose 3: Multiplication")
            a = float(input("Enter the 1st factor: ").strip())
            b = float(input("Enter the 2nd factor: ").strip())
            result = multiply(a, b)
        elif operator == 4:
            print("-------------------------------------------")
            print("you choose 4: Division")
            a = float(input("Enter the dividend: ").strip())
            while True:
                b = float(input("Enter the divisor: ").strip())
                if b != 0:
                    break
                print("Divisor cannot be 0. Please enter a divisor again.")
            result = divide(a, b)
        elif operator == 5:
            print("-------------------------------------------")
            print("you choose 5: Exponentiation")
            a = float(input("Enter the base: ").strip())
            b = float(input("Enter the exponent: ").strip())
            result = power(a, b)

        print("-------------------------------------------")
        print(f"Your answer is {result:.2f}")
        print("-------------------------------------------")
        cont = input("Do you want to exit (Y / N): ")
        if (cont == 'Y') or (cont == 'y'):
            print("-------------------------------------------")
            print("Thank you...")
            break
    except (ValueError):
        print("-------------------------------------------")
        print("Please Enter number")
```

### Lab 6.2 : Custom Module and Standard Library Usage

```python
%%writefile my_utils.py

def greet(name):
    print(f"Hello, {name}!")

def is_prime(number):
    if number <= 1:
        return False
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            return False
    return True
```

```python
import my_utils
import math
import random

my_utils.greet("Alice")
print("Is 7 prime?:", my_utils.is_prime(7))
print("Is 10 prime?:", my_utils.is_prime(10))
print("Square root of 16:", math.sqrt(16))
print("Random number:", random.randint(1, 100))
```
