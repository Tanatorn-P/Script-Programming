### Activity 6.1: Refactoring a Previous Lab Assignment

```python
# Simple Guessing Game จาก Lab 3.2
import random

def get_user_input(prompt: str) -> int | None:
    """รับค่าตัวเลขจากผู้ใช้ พร้อมจัดการข้อผิดพลาดกรณีใส่ข้อมูลไม่ถูกต้อง"""
    try:
        return int(input(prompt))
    except ValueError:
        return None

def check_guess(guess: int, target: int) -> str:
    """เปรียบเทียบตัวเลขที่ทายกับค่าเป้าหมาย แล้วส่งคืนสถานะ"""
    if guess == target:
        return "CORRECT"
    elif guess < target:
        return "LOW"
    else:
        return "HIGH"

def play_guessing_game(max_attempts: int = 5, min_val: int = 1, max_val: int = 50) -> bool:
    """ฟังก์ชันหลักสำหรับควบคุมการดำเนินเกม"""
    rand_number = random.randint(min_val, max_val)
    attempts = 0

    while attempts < max_attempts:
        number = get_user_input(f"Guess a number {min_val} - {max_val} : ")

        # กรณีป้อนข้อมูลไม่ใช่ตัวเลข
        if number is None:
            attempts += 1
            remaining = max_attempts - attempts
            print(f"--ใส่ตัวเลข บักปอบ ({remaining} times left)\n")
            continue

        # ตรวจสอบผลการทาย
        result = check_guess(number, rand_number)

        if result == "CORRECT":
            print("Congratulations! You guessed it!")
            return True

        # กรณีทายผิด
        attempts += 1
        remaining = max_attempts - attempts
        hint = "too low" if result == "LOW" else "too high"
        print(f"--{number} is {hint}! Try again. ({remaining} times left)\n")

    # สรุปผลเมื่อใช้สิทธิ์ครบ
    print(f"--Sorry you lose. You've guessed for {max_attempts} times!")
    print(f"--The correct number is : {rand_number}")
    return False

# เรียกใช้งานเกม
if __name__ == "__main__":
    play_guessing_game()
```
