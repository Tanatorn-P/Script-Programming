## [Version 1.0.0]

### 1. Computer Vision & OCR Optimization

> **หัวข้อการเรียนรู้:** การประมวลผลภาพถ่ายสเกลบาร์และการเพิ่มความแม่นยำในการอ่านตัวเลขด้วย OCR

* **Targeted Region Cropping (การจำกัดพื้นที่ประมวลผล)**
* **บทเรียนที่ได้:** การส่งภาพถ่าย SEM ทั้งใบเข้า OCR ทำให้ประมวลผลช้าและเกิด Noise จากข้อความอื่นบนภาพ
* **การแก้ไข:** กำหนดพิกัด Bounding Box `CROP_BOX = (453, 908, 514, 930)` เจาะจงเฉพาะแถบสเกลบาร์ ช่วยลดเวลาการประมวลผลของ EasyOCR ลงกว่า **80%**

### 2. Data Cleaning & Type Normalization

> **หัวข้อการเรียนรู้:** การทำความสะอาดข้อมูลและการปรับมาตรฐานรูปแบบข้อมูลระหว่าง Excel กับระบบไฟล์

* **Merged Cells Handling (การจัดการเซลล์ผสาน)**
* **บทเรียนที่ได้:** ไฟล์ Excel จากผู้ใช้มักมีการผสานเซลล์ (Merged Cells) ในคอลัมน์วันที่ ทำให้แถวถัดมารับค่าเป็น `NaN`
* **การแก้ไข:** ใช้คำสั่ง `df['Date'].ffill()` (Forward Fill) เพื่อเติมค่าว่างอัตโนมัติก่อนทำ Data Mapping

### 3. Software Safety & Architecture Design

> **หัวข้อการเรียนรู้:** สถาปัตยกรรมระบบ ความปลอดภัยไฟล์ และการรับมือกับข้อผิดพลาด (Fault Tolerance)

* **DRY_RUN Safety Execution Pattern**
* **บทเรียนที่ได้:** การสั่งเปลี่ยนชื่อไฟล์จำนวนมากในระบบมีโอกาสเกิดความผิดพลาดสูง
* **การแก้ไข:** กำหนดโหมด `dry_run=True` เป็นค่าเริ่มต้น เพื่อพรีวิวชื่อไฟล์ใหม่บนหน้าจอก่อนลงมือเปลี่ยนชื่อไฟล์จริง

## [Version 1.3.0]

### 1. Computer Vision & OCR Optimization

> **หัวข้อการเรียนรู้:** การประมวลผลภาพถ่ายสเกลบาร์และการเพิ่มความแม่นยำในการอ่านตัวเลขด้วย OCR

* **Multi-Stage Preprocessing Pipeline (การปรุงแต่งภาพก่อนทำ OCR)**
* **บทเรียนที่ได้:** ตัวเลขกำลังขยายบนสเกลบาร์ SEM มักมีความคมชัดต่ำ (Low Contrast) ส่งผลให้ OCR อ่านผิดพลาด
* **การพัฒนาต่อยอดจาก v1.0.0:** สร้าง OpenCV Pipeline เต็มรูปแบบ:
1. Grayscale Conversion
2. Resize ขยาย 3 เท่าแบบ Cubic Interpolation
3. Binary Thresholding (Cutoff = 200)
4. Erosion (Kernel $2 \times 2$) เพื่อเชื่อมเส้นอักษรที่ขาดออกจากกัน


* **ผลลัพธ์:** เพิ่มความแม่นยำในการอ่านค่ากำลังขยายขึ้นเกิน **90%**

### 2. Algorithm & Sequential State Logic

> **หัวข้อการเรียนรู้:** อัลกอริทึมการวิเคราะห์ลำดับภาพถ่ายอัตโนมัติจากกำลังขยาย

* **Dynamic Sequential State Logic (ตรรกะเปรียบเทียบกำลังขยาย)**
* **บทเรียนที่ได้:** การถ่ายภาพ SEM ของหินแต่ละก้อนจะมีการซูมขยายหลายระดับ ณ จุดเดิม ก่อนย้ายไปจุดถ่ายใหม่
* **การพัฒนาต่อยอดจาก v1.1.0:** พัฒนา State Logic เปรียบเทียบกำลังขยายภาพปัจจุบัน ($mag_{curr}$) กับภาพถัดไป ($mag_{next}$):
* หาก $mag_{curr} < mag_{next} \rightarrow$ **จุดเดิมแต่ซูมลึกขึ้น** (เพิ่ม `zoom_num += 1`)
* หาก $mag_{curr} \ge mag_{next} \rightarrow$ **ย้ายตำแหน่งจุดถ่ายใหม่** (เพิ่ม `point_num += 1` และรีเซ็ต `zoom_num = 1`)
