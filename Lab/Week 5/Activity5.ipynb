### กิจกรรมที่ 5.1: การประยุกต์ใช้ Dictionary และ Set ในชีวิตจริง

####**คำถามกระตุ้นความคิด (Prompts):**
1. คุณจะนำ **Dictionary** ไปใช้ตรงไหนในเกมได้บ้าง? *(เช่น ช่องเก็บของของผู้เล่น (Inventory), ค่าพลังของตัวละคร)*<br>
ตอบ
- Inventory System: ใช้ ID ไอเทมเป็น Key และจำนวนชิ้นเป็น Value เช่น {"potion": 5, "coin": 100}
- Character Status: เก็บค่าพลังตัวละคร เช่น {"hp": 100, "atk": 25, "def": 10}<br><br>

2. **Dictionary** จะช่วยจัดเก็บข้อมูลสินค้าในร้านค้าออนไลน์ได้อย่างไร? *(เช่น ข้อมูลรายละเอียดสินค้าโดยใช้อ้างอิงจากรหัสสินค้า/ID)*<br>
ตอบ **Product Catalog :** ใช้ Product ID เป็น Key เพื่อค้นหาข้อมูลสินค้า (ชื่อ, ราคา, สต็อก) ได้ทันทีโดยไม่ต้องวนลูปหา เช่น {"P101": {"name": "Shirt", "price": 350}}<br><br>
3. **Set** จะมีประโยชน์อย่างไรในการจัดการสิทธิ์ของผู้ใช้ หรือการจัดการแท็ก? *(เช่น บทบาทสิทธิ์ใช้งานที่ไม่ซ้ำกัน, แฮชแท็กที่ไม่ซ้ำกัน)*<br>
ตอบ
- สิทธิ์ผู้ใช้ (Roles/Permissions): ป้องกันไม่ให้รับสิทธิ์ซ้ำซ้อน และใช้เปรียบเทียบสิทธิ์ (Union/Intersection) ได้ง่าย
- แฮชแท็ก (Hashtags): ตัดแท็กที่ซ้ำกันออกโดยอัตโนมัติ เพื่อให้ได้กลุ่มคำค้นหาที่ไม่ซ้ำกัน<br><br>

4. นึกออกไหมว่ามีสถานการณ์ไหนบ้างที่เราต้องเช็กอย่างรวดเร็วว่า **"มีข้อมูลนี้อยู่ในกลุ่มข้อมูลขนาดใหญ่หรือไม่"** โดยไม่สนใจลำดับก่อนหลัง? *(เช่น รายชื่อ IP Address ที่ถูกบล็อก, การนับจำนวนผู้เข้าชมเว็บที่ไม่นับคนซ้ำ)*<br>
ตอบ
- IP Blacklist: เช็กทันทีว่า IP ที่กำลังเชื่อมต่อเข้ามาอยู่ในรายชื่อที่ถูกบล็อกหรือไม่
- Unique Visitors (UV): เช็กและนับจำนวนผู้ใช้งานแบบไม่นับคนซ้ำ (Unique User ID)

### กิจกรรมที่ 5.2: โปรแกรมนับความถี่ของคำ (Word Frequency Counter) โดยใช้ Dictionary

```python
sentence = input("Enter a sentence: ")
sentence_clean = sentence.lower()

words = sentence_clean.split() #แยกคำออกจากกัน
word_counts = {}
for word in words:
    clean_word = word.strip('.,!?"():') #ลบเครื่องหมายวรรคตอนของคำ
    if not clean_word:
        continue
    if clean_word in word_counts: #เข็คว่ามีคำอยู่ใน dictiionary หรือยัง
        word_counts[clean_word] += 1 #ถ้ามีแล้ว ให้เพิ่มค่าจำนวนนับขึ้นอีก 1
    else:
        word_counts[clean_word] = 1 #ถ้ายังไม่มี ให้กำหนดค่าเริ่มต้นเป็น 1
print("\n--- Word Counts ---")
print(f"{'WORD':<15} | {'COUNT':<5}")
print("-" * 23)

for word, count in word_counts.items():
    print(f"{word:<15} | {count:<5}")
```
